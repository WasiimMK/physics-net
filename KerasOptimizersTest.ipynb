```python
import tensorflow as tf
import matplotlib.pyplot as plt
import numpy as np
```

```python
f = lambda x, y : (x + 2*y - 7)**2 + (2*x + y - 5)**2

x = np.linspace(0, 4, 100)
y = np.linspace(0, 4, 100)
```

```python
X, Y = np.meshgrid(x, y, indexing = 'xy')

print(X.shape)
print(Y.shape)
```

```python
plt.contourf(X, Y, f(X, Y), cmap=plt.cm.coolwarm)
```

```python
fig = plt.figure()
ax = plt.axes(projection='3d')
ax.contour3D(X, Y, f(X, Y), 50, cmap='hsv')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
```

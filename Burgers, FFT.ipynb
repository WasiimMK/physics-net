```python
import numpy as np
import math
import matplotlib.pyplot as plt
import matplotlib.cm as cm
from scipy.integrate import odeint
import scipy.linalg as la
import time
```

```python
nu = 0.00001

t0 = 0
tmax = 1

x0 = 0
xmax = 10

t_step = 0.01
x_step = 0.001

N_t = int(((tmax - t0)/t_step) + 1)
N_x = int(((xmax - x0)/x_step) + 1)

t = np.linspace(t0, tmax, N_t)
x = np.linspace(x0, xmax, N_x)

N_t = len(t)
N_x = len(x)

dt = t[1] - t[0]
dx = x[1] - x[0]

print(N_t, dt)
print(N_x, dx)
```

```python
def f(x) :
        
    return np.sin((2 * np.pi * x)/xmax)
```

```python
kappa = 2*np.pi*np.fft.fftfreq(N_x, dx)

u0 = f(x)
```

```python
def rhsBurgers(u, t, kappa, nu) :

    uhat = np.fft.fft(u)
    d_uhat_dx = (1j) * kappa * uhat
    d2_uhat_dx2 = -(kappa ** 2) * uhat
    d_u_dx = np.fft.ifft(d_uhat_dx)
    d2_u_dx2 = np.fft.ifft(d2_uhat_dx2)
    d_u_dt = (nu * d2_u_dx2) - (u * d_u_dx)
    
    d_u_dt_real = d_u_dt.real
    
    return d_u_dt_real
```

```python
start = time.time()
```

```python
u = odeint(rhsBurgers, u0, t, (kappa, nu))
```

```python
fig = plt.figure()
ax = fig.add_subplot(111, projection = '3d')
ax.set_xlabel("x")
ax.set_ylabel("t")
ax.set_zlabel("u")


u_plot = u[0:-1:10, :]

for j in range(u_plot.shape[0]) :

    ys = j * np.ones(u_plot.shape[1])
    ax.plot(x, ys, u_plot[j,:], color = cm.jet(j*20))
    
plt.figure()
plt.xlabel("x")
plt.ylabel("t")
plt.imshow(np.flipud(u), aspect = 8, interpolation = 'none', extent = [0, xmax, 0, tmax])
plt.set_cmap('jet_r')
plt.colorbar()
```
